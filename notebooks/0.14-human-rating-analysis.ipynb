```python
import pandas as pd
import numpy as np

df = pd.read_csv("../human_ratings/HumanRatings.csv")
df.columns = df.columns.str.strip()

trial_cols = ["Trial 1", "Trial 2", "Trial 3", "Trial 4", "Trial 5", "Trial 6"]
ratings = df[trial_cols].values  # (n_samples, 6)

# Leave-one-out ground truth: mean of the other 5 raters for each rating
n, k = ratings.shape
row_sums = ratings.sum(axis=1, keepdims=True)
loo_means = (row_sums - ratings) / (k - 1)  # (n_samples, 6)

diffs = np.abs(ratings - loo_means)

total = diffs.size
for threshold in [1, 2, 3, 4]:
    pct = (diffs <= threshold).sum() / total * 100
    print(f"Within ±{threshold}: {pct:.2f}%")

avg_std = df["std"].mean()
print(f"\nAverage panel std: {avg_std:.3f}")
```
